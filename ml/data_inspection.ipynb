```python
import pandas as pd
import numpy as np

# Load dataset
df = pd.read_csv("../ml/data/tech_mental_health_burnout.csv")

# --------------------------------------------------
# FEATURES AND TARGET
# --------------------------------------------------

# Remove both burnout_score and burnout_level
X = df.drop(columns=["burnout_score", "burnout_level"])

# Target
y = df["burnout_score"]

print("X shape:", X.shape)
print("y shape:", y.shape)
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("Training data:", X_train.shape)
print("Testing data :", X_test.shape)
```

```python
categorical_columns = [
    "gender",
    "job_role",
    "company_size",
    "work_mode"
]

numerical_columns = [
    "age",
    "experience_years",
    "work_hours_per_week",
    "overtime_hours",
    "meetings_per_day",
    "deadlines_missed",
    "job_satisfaction",
    "manager_support",
    "work_life_balance",
    "sleep_hours",
    "physical_activity_days",
    "screen_time_hours",
    "caffeine_intake",
    "social_support_score",
    "has_therapy",
    "stress_level",
    "anxiety_score",
    "depression_score",
    "seeks_professional_help"
]
```

# **Preprocessing Pipeline**

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder

preprocessor = ColumnTransformer(
    transformers=[
        (
            "categorical",
            OneHotEncoder(handle_unknown="ignore"),
            categorical_columns
        ),
        (
            "numerical",
            "passthrough",
            numerical_columns
        )
    ]
)
```

```python
X_train_processed = preprocessor.fit_transform(X_train)

X_test_processed = preprocessor.transform(X_test)

print("Processed training shape:", X_train_processed.shape)
print("Processed testing shape :", X_test_processed.shape)
```

# **Train Linear Regression**

```python
from sklearn.linear_model import LinearRegression

linear_model = LinearRegression()

linear_model.fit(
    X_train_processed,
    y_train
)

linear_predictions = linear_model.predict(X_test_processed)
print("Linear Regression is trained!")
```

```python
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

linear_mae = mean_absolute_error(
    y_test,
    linear_predictions
)

linear_mse = mean_squared_error(
    y_test,
    linear_predictions
)

linear_rmse = np.sqrt(linear_mse)

linear_r2 = r2_score(
    y_test,
    linear_predictions
)

print("Linear Regression")
print("-------------------------")
print("MAE :", linear_mae)
print("MSE :", linear_mse)
print("RMSE:", linear_rmse)
print("R²  :", linear_r2)
```

# **Train Random Forest**

```python
from sklearn.ensemble import RandomForestRegressor

rf_model = RandomForestRegressor(
    n_estimators=100,
    random_state=42,
    n_jobs=-1
)

rf_model.fit(
    X_train_processed,
    y_train
)


rf_predictions = rf_model.predict(X_test_processed)
print("random forest Trained!")
```

```python
rf_mae = mean_absolute_error(
    y_test,
    rf_predictions
)

rf_mse = mean_squared_error(
    y_test,
    rf_predictions
)

rf_rmse = np.sqrt(rf_mse)

rf_r2 = r2_score(
    y_test,
    rf_predictions
)

print("Random Forest")
print("-------------------------")
print("MAE :", rf_mae)
print("MSE :", rf_mse)
print("RMSE:", rf_rmse)
print("R²  :", rf_r2)
```

# **Train XGBoost Model**

```python
from xgboost import XGBRegressor

xgb_model = XGBRegressor(
    n_estimators=300,
    learning_rate=0.05,
    max_depth=6,
    subsample=0.8,
    colsample_bytree=0.8,
    objective="reg:squarederror",
    random_state=42,
    n_jobs=-1
)

xgb_model.fit(
    X_train_processed,
    y_train
)

xgb_predictions = xgb_model.predict(X_test_processed)
print("XGBoost is Trained!")
```

```python
xgb_mae = mean_absolute_error(
    y_test,
    xgb_predictions
)

xgb_mse = mean_squared_error(
    y_test,
    xgb_predictions
)

xgb_rmse = np.sqrt(xgb_mse)

xgb_r2 = r2_score(
    y_test,
    xgb_predictions
)

print("XGBoost")
print("-------------------------")
print("MAE :", xgb_mae)
print("MSE :", xgb_mse)
print("RMSE:", xgb_rmse)
print("R²  :", xgb_r2)
```

```python
# ============================================================
# STEP 1: PERFORMANCE BY BURNOUT LEVEL
# ============================================================

from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np
import pandas as pd

# Create results dataframe
level_results = X_test.copy()

level_results["actual_score"] = y_test.values
level_results["predicted_score"] = xgb_predictions

# Get actual burnout level from original dataframe
level_results["burnout_level"] = df.loc[X_test.index, "burnout_level"].values

# Calculate metrics for each burnout level
results = []

for level in ["Low", "Moderate", "High"]:

    data = level_results[level_results["burnout_level"] == level]

    mae = mean_absolute_error(
        data["actual_score"],
        data["predicted_score"]
    )

    rmse = np.sqrt(
        mean_squared_error(
            data["actual_score"],
            data["predicted_score"]
        )
    )

    r2 = r2_score(
        data["actual_score"],
        data["predicted_score"]
    ) if len(data) > 1 else np.nan

    results.append({
        "Burnout Level": level,
        "Cases": len(data),
        "MAE": mae,
        "RMSE": rmse,
        "R2": r2
    })

level_performance = pd.DataFrame(results)

print("\nPerformance by Burnout Level")
print("=" * 60)
print(level_performance.to_string(index=False))
```

```python
# ============================================================
# STEP 2: INSPECT MODERATE AND HIGH PREDICTIONS
# ============================================================

# Moderate cases
moderate_cases = level_results[
    level_results["burnout_level"] == "Moderate"
].copy()

moderate_cases["error"] = (
    moderate_cases["actual_score"]
    - moderate_cases["predicted_score"]
)

moderate_cases["absolute_error"] = (
    moderate_cases["error"].abs()
)

print("\nModerate Burnout Cases")
print("=" * 70)

print(
    moderate_cases[
        [
            "actual_score",
            "predicted_score",
            "error",
            "absolute_error"
        ]
    ].head(20).to_string(index=False)
)


# High cases
high_cases = level_results[
    level_results["burnout_level"] == "High"
].copy()

high_cases["error"] = (
    high_cases["actual_score"]
    - high_cases["predicted_score"]
)

high_cases["absolute_error"] = (
    high_cases["error"].abs()
)

print("\n\nHigh Burnout Cases")
print("=" * 70)

print(
    high_cases[
        [
            "actual_score",
            "predicted_score",
            "error",
            "absolute_error"
        ]
    ].to_string(index=False)
)
```

```python
# ============================================================
# STEP 4: UNDERPREDICTION CHECK
# ============================================================

comparison = pd.DataFrame({
    "Actual": y_test.values,
    "Predicted": xgb_predictions
})

comparison["Error"] = (
    comparison["Actual"] - comparison["Predicted"]
)

comparison["Absolute_Error"] = (
    comparison["Error"].abs()
)

print("\nOverall Error Bias")
print("=" * 60)

print(
    "Mean Error:",
    comparison["Error"].mean()
)

print(
    "Mean Absolute Error:",
    comparison["Absolute_Error"].mean()
)

print(
    "Underpredictions:",
    (comparison["Error"] > 0).sum()
)

print(
    "Overpredictions:",
    (comparison["Error"] < 0).sum()
)
```

```python
# ============================================================
# STEP 5: XGBOOST HYPERPARAMETER TUNING
# ============================================================

from xgboost import XGBRegressor
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)
import numpy as np
import pandas as pd
import time


# ------------------------------------------------------------
# Models to test
# ------------------------------------------------------------

models = {

    "XGB_Current": XGBRegressor(
        n_estimators=300,
        learning_rate=0.05,
        max_depth=6,
        subsample=0.8,
        colsample_bytree=0.8,
        objective="reg:squarederror",
        random_state=42,
        n_jobs=-1
    ),

    "XGB_Tuned_1": XGBRegressor(
        n_estimators=500,
        learning_rate=0.03,
        max_depth=6,
        min_child_weight=3,
        subsample=0.85,
        colsample_bytree=0.85,
        objective="reg:squarederror",
        random_state=42,
        n_jobs=-1
    ),

    "XGB_Tuned_2": XGBRegressor(
        n_estimators=500,
        learning_rate=0.03,
        max_depth=8,
        min_child_weight=2,
        subsample=0.85,
        colsample_bytree=0.85,
        objective="reg:squarederror",
        random_state=42,
        n_jobs=-1
    ),

    "XGB_Tuned_3": XGBRegressor(
        n_estimators=400,
        learning_rate=0.04,
        max_depth=7,
        min_child_weight=1,
        subsample=0.9,
        colsample_bytree=0.9,
        objective="reg:squarederror",
        random_state=42,
        n_jobs=-1
    )
}


# ------------------------------------------------------------
# Train and evaluate
# ------------------------------------------------------------

tuning_results = []

trained_models = {}

for name, model in models.items():

    print("\nTraining:", name)

    start_time = time.time()

    model.fit(
        X_train_processed,
        y_train
    )

    predictions = model.predict(
        X_test_processed
    )

    mae = mean_absolute_error(
        y_test,
        predictions
    )

    mse = mean_squared_error(
        y_test,
        predictions
    )

    rmse = np.sqrt(mse)

    r2 = r2_score(
        y_test,
        predictions
    )

    training_time = time.time() - start_time

    tuning_results.append({
        "Model": name,
        "MAE": mae,
        "MSE": mse,
        "RMSE": rmse,
        "R2": r2,
        "Training_Time_sec": training_time
    })

    trained_models[name] = model

    print("MAE :", round(mae, 6))
    print("RMSE:", round(rmse, 6))
    print("R2  :", round(r2, 6))
    print("Time:", round(training_time, 2), "seconds")


# ------------------------------------------------------------
# Comparison table
# ------------------------------------------------------------

tuning_results_df = pd.DataFrame(
    tuning_results
)

print("\n")
print("=" * 80)
print("XGBOOST TUNING RESULTS")
print("=" * 80)

print(
    tuning_results_df
    .sort_values("R2", ascending=False)
    .to_string(index=False)
)
```

```python
# ============================================================
# STEP 6: TEST TUNED MODELS BY BURNOUT LEVEL
# ============================================================

from sklearn.metrics import mean_absolute_error, mean_squared_error
import numpy as np
import pandas as pd

level_tuning_results = []

for name, model in trained_models.items():

    predictions = model.predict(X_test_processed)

    temp = pd.DataFrame({
        "actual": y_test.values,
        "predicted": predictions,
        "level": df.loc[X_test.index, "burnout_level"].values
    })

    for level in ["Low", "Moderate", "High"]:

        data = temp[temp["level"] == level]

        mae = mean_absolute_error(
            data["actual"],
            data["predicted"]
        )

        rmse = np.sqrt(
            mean_squared_error(
                data["actual"],
                data["predicted"]
            )
        )

        level_tuning_results.append({
            "Model": name,
            "Level": level,
            "Cases": len(data),
            "MAE": mae,
            "RMSE": rmse
        })


level_tuning_df = pd.DataFrame(level_tuning_results)

print("\n")
print("=" * 90)
print("XGBOOST PERFORMANCE BY BURNOUT LEVEL")
print("=" * 90)

print(
    level_tuning_df.to_string(index=False)
)
```

```python
# ============================================================
# STEP 7: COMPARE HIGH-BURNOUT PREDICTIONS
# ============================================================

high_comparison = pd.DataFrame({
    "Actual": y_test.values,
    "Current": trained_models["XGB_Current"].predict(X_test_processed),
    "Tuned_1": trained_models["XGB_Tuned_1"].predict(X_test_processed),
    "Tuned_2": trained_models["XGB_Tuned_2"].predict(X_test_processed),
    "Tuned_3": trained_models["XGB_Tuned_3"].predict(X_test_processed),
    "Level": df.loc[X_test.index, "burnout_level"].values
})

# Keep only High cases
high_comparison = high_comparison[
    high_comparison["Level"] == "High"
].copy()

print("\n")
print("=" * 90)
print("HIGH BURNOUT PREDICTION COMPARISON")
print("=" * 90)

print(
    high_comparison[
        ["Actual", "Current", "Tuned_1", "Tuned_2", "Tuned_3"]
    ].to_string(index=False)
)
```

```python
# ============================================================
# STEP 8: SELECT FINAL MODEL CANDIDATE
# ============================================================

final_model = trained_models["XGB_Tuned_1"]

print("Selected model: XGB_Tuned_1")
print(final_model)
```

```python
# ============================================================
# FINAL MODEL EVALUATION
# ============================================================

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)
import numpy as np

final_predictions = final_model.predict(
    X_test_processed
)

final_mae = mean_absolute_error(
    y_test,
    final_predictions
)

final_mse = mean_squared_error(
    y_test,
    final_predictions
)

final_rmse = np.sqrt(final_mse)

final_r2 = r2_score(
    y_test,
    final_predictions
)

print("\nFINAL XGBOOST MODEL PERFORMANCE")
print("=" * 50)

print(f"MAE  : {final_mae:.6f}")
print(f"MSE  : {final_mse:.6f}")
print(f"RMSE : {final_rmse:.6f}")
print(f"R2   : {final_r2:.6f}")
```

```python
# ============================================================
# STEP 9: SAVE FINAL MODEL AND PREPROCESSOR
# ============================================================

import joblib
from pathlib import Path

# Create model directory
MODEL_DIR = Path("model")
MODEL_DIR.mkdir(exist_ok=True)

# Save trained XGBoost model
joblib.dump(
    final_model,
    MODEL_DIR / "xgb_burnout_model.pkl"
)

# Save fitted preprocessor
joblib.dump(
    preprocessor,
    MODEL_DIR / "preprocessor.pkl"
)

# Save feature names
joblib.dump(
    list(X_train.columns),
    MODEL_DIR / "feature_names.pkl"
)

print("Model saved successfully!")
print("Saved files:")

for file in MODEL_DIR.iterdir():
    print(file)
```

```python
import joblib
import shap
import pandas as pd
from pathlib import Path

# Load saved files
MODEL_DIR = Path("model")

final_model = joblib.load(
    MODEL_DIR / "xgb_burnout_model.pkl"
)

preprocessor = joblib.load(
    MODEL_DIR / "preprocessor.pkl"
)

feature_names = joblib.load(
    MODEL_DIR / "feature_names.pkl"
)

print("Model loaded successfully!")
print("Number of input features:", len(feature_names))
```

```python
# Create SHAP explainer
explainer = shap.TreeExplainer(final_model)

print("SHAP explainer created successfully!")
```

```python
# Take 100 samples from the test dataset
X_sample = X_test.iloc[:100]

# Apply the same preprocessing used during training
X_sample_processed = preprocessor.transform(X_sample)

# Calculate SHAP values
shap_values = explainer.shap_values(
    X_sample_processed
)

print("SHAP values calculated successfully!")
print("SHAP shape:", shap_values.shape)
```

```python
shap.summary_plot(
    shap_values,
    X_sample_processed,
    feature_names=preprocessor.get_feature_names_out()
)
```

```python
shap.summary_plot(
    shap_values,
    X_sample_processed,
    feature_names=preprocessor.get_feature_names_out(),
    plot_type="bar"
)
```
