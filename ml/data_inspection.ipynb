## **Preprocess 30K dataset**

```python
import pandas as pd

df = pd.read_csv("../ml/data/burnout_30k.csv")

print("Shape:", df.shape)
print(df.head())
```

```python
X = df.drop(columns=["burnout_level", "burnout_score"])
y = df["burnout_level"]

print("X shape:", X.shape)
print("y shape:", y.shape)
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training:", X_train.shape)
print("Testing :", X_test.shape)
```

```python
print("Training distribution:")
print(y_train.value_counts())

print("\nTesting distribution:")
print(y_test.value_counts())
```

```python
categorical_features = [
    "gender",
    "job_role",
    "company_size",
    "work_mode"
]

numerical_features = [
    "age",
    "experience_years",
    "work_hours_per_week",
    "overtime_hours",
    "meetings_per_day",
    "deadlines_missed",
    "job_satisfaction",
    "manager_support",
    "work_life_balance",
    "sleep_hours",
    "physical_activity_days",
    "screen_time_hours",
    "caffeine_intake",
    "social_support_score",
    "has_therapy",
    "stress_level",
    "anxiety_score",
    "depression_score",
    "seeks_professional_help"
]

print("Categorical features:", len(categorical_features))
print("Numerical features:", len(numerical_features))
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder

preprocessor = ColumnTransformer(
    transformers=[
        (
            "categorical",
            OneHotEncoder(handle_unknown="ignore"),
            categorical_features
        ),
        (
            "numerical",
            "passthrough",
            numerical_features
        )
    ]
)

print("Preprocessing pipeline created successfully.")
```

```python
X_train_processed = preprocessor.fit_transform(X_train)

X_test_processed = preprocessor.transform(X_test)

print("Processed training shape:", X_train_processed.shape)
print("Processed testing shape:", X_test_processed.shape)
```

```python
print("Original features:", X_train.shape[1])
print("Processed features:", X_train_processed.shape[1])
```

```python
print("Training class distribution:")
print(y_train.value_counts())

print("\nTraining class percentages:")
print(y_train.value_counts(normalize=True) * 100)
```

```python
# Target
y = df["burnout_score"]

# Features
X = df.drop(columns=["burnout_score", "burnout_level"])

print("X shape:", X.shape)
print("y shape:", y.shape)

print("\nFeatures:")
print(X.columns.tolist())
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("Training data:", X_train.shape)
print("Testing data:", X_test.shape)
```

```python
X_train_processed = preprocessor.fit_transform(X_train)
X_test_processed = preprocessor.transform(X_test)
```

```python
print("Original features:", X_train.shape[1])
print("Processed features:", X_train_processed.shape[1])
```

# **Baseline Model (Linear Regression)**

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
from sklearn.linear_model import LinearRegression

linear_model = LinearRegression()

linear_model.fit(
    X_train_processed,
    y_train
)

y_pred_linear = linear_model.predict(X_test_processed)

print("Linear Regression trained successfully.")
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

mae = mean_absolute_error(y_test, y_pred_linear)
mse = mean_squared_error(y_test, y_pred_linear)
rmse = np.sqrt(mse)
r2 = r2_score(y_test, y_pred_linear)

print("Linear Regression Results")
print("-------------------------")
print("MAE :", mae)
print("MSE :", mse)
print("RMSE:", rmse)
print("R²  :", r2)
```

# **Comparison Model Random Forest Train**

```python
from sklearn.ensemble import RandomForestRegressor

rf_model = RandomForestRegressor(
    n_estimators=200,
    random_state=42,
    n_jobs=-1
)

rf_model.fit(X_train_processed, y_train)

y_pred_rf = rf_model.predict(X_test_processed)

print("Model Random Forest is trained!")
```

# **Comparison Model XGBoost Train**

```python
from xgboost import XGBRegressor

xgb_model = XGBRegressor(
    n_estimators=200,
    learning_rate=0.05,
    max_depth=6,
    random_state=42
)

xgb_model.fit(X_train_processed, y_train)

y_pred_xgb = xgb_model.predict(X_test_processed)
print("Model XGBoost is trained!")
```

## **To check Accuracy compare 3 Models**

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np
import pandas as pd

# Predictions
y_pred_linear = linear_model.predict(X_test_processed)
y_pred_rf = rf_model.predict(X_test_processed)
y_pred_xgb = xgb_model.predict(X_test_processed)


# Evaluation function
def evaluate_model(name, y_test, y_pred):

    mae = mean_absolute_error(y_test, y_pred)
    mse = mean_squared_error(y_test, y_pred)
    rmse = np.sqrt(mse)
    r2 = r2_score(y_test, y_pred)

    return {
        "Model": name,
        "MAE": mae,
        "MSE": mse,
        "RMSE": rmse,
        "R2 Score": r2
    }


# Evaluate all models
results = []

results.append(
    evaluate_model("Linear Regression", y_test, y_pred_linear)
)

results.append(
    evaluate_model("Random Forest", y_test, y_pred_rf)
)

results.append(
    evaluate_model("XGBoost", y_test, y_pred_xgb)
)


# Create comparison table
results_df = pd.DataFrame(results)

print(results_df)
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Get feature names after preprocessing
feature_names = preprocessor.get_feature_names_out()

# Get XGBoost feature importance
importance = xgb_model.feature_importances_

# Create dataframe
feature_importance_df = pd.DataFrame({
    "Feature": feature_names,
    "Importance": importance
})

# Sort by importance
feature_importance_df = feature_importance_df.sort_values(
    by="Importance",
    ascending=False
)

print(feature_importance_df.head(15))
```
