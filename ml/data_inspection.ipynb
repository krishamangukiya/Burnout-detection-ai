# **Reduce 150K TO 30K Rows**

```python
# Keep ALL High-risk records
high = df[df["burnout_level"] == "High"]

# Sample Moderate records
moderate = df[df["burnout_level"] == "Moderate"].sample(
    n=8000,
    random_state=42
)

# Sample Low records
low = df[df["burnout_level"] == "Low"].sample(
    n=21935,
    random_state=42
)

# Combine
df_30k = pd.concat([low, moderate, high], ignore_index=True)

# Shuffle
df_30k = df_30k.sample(
    frac=1,
    random_state=42
).reset_index(drop=True)

print(df_30k.shape)
print(df_30k["burnout_level"].value_counts())
```

```python
print("Dataset Shape:", df_30k.shape)
```

```python
print("\nBurnout Distribution:")
print(df_30k["burnout_level"].value_counts())
```

```python
print("\nBurnout Distribution (%):")
print(df_30k["burnout_level"].value_counts(normalize=True) * 100)
```

```python
print("Missing values:", df_30k.isnull().sum().sum())
```

```python
print("Duplicate rows:", df_30k.duplicated().sum())
```

```python
X = df_30k.drop(columns=["burnout_level", "burnout_score"])

y = df_30k["burnout_level"]

print("X shape:", X.shape)
print("y shape:", y.shape)
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training data:", X_train.shape)
print("Testing data:", X_test.shape)
```

```python
print("\nTraining distribution:")
print(y_train.value_counts())

print("\nTesting distribution:")
print(y_test.value_counts())
```
